#(MEDIA-MEDIANA-MODA)

```python
#MEDIA
mean_income = df['Income'].mean()
mean_age = (2023 - df['Year_Birth']).mean()
mean_spent = df[['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']].mean().mean()
mean_kids = df[['Kidhome', 'Teenhome']].mean().mean()

print(f"Media de Ingresos: {mean_income}")
print(f"Media de Edad: {mean_age}")
print(f"Media de Cantidad Gastada en Productos: {mean_spent}")
print(f"Media de Número de Hijos: {mean_kids}")


```

```python
#MEDIANA
median_income = df['Income'].median()
median_age = (2023 - df['Year_Birth']).median()
median_spent = df[['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']].median().median()
median_kids = df[['Kidhome', 'Teenhome']].median().median()

print(f"Mediana de Ingresos: {median_income}")
print(f"Mediana de Edad: {median_age}")
print(f"Mediana de Cantidad Gastada en Productos: {median_spent}")
print(f"Mediana de Número de Hijos: {median_kids}")
```

```python
# MODA
mode_income = df['Income'].mode()[0]
mode_age = (2023 - df['Year_Birth']).mode()[0]
mode_spent = df[['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']].mode().iloc[0].mean()
mode_kids = df[['Kidhome', 'Teenhome']].mode().iloc[0].mean()

print(f"Moda de Ingresos: {mode_income}")
print(f"Moda de Edad: {mode_age}")
print(f"Moda de Cantidad Gastada en Productos (promedio de vino, frutas, carne, etc.): {mode_spent}")
print(f"Moda de Número de Hijos: {mode_kids}")
```

#Resumen estadistico de los datos obtenidos

```python
# Resumen estadístico
income_summary = df['Income'].describe()
age_summary = (2023 - df['Year_Birth']).describe()
spent_summary = df[['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']].describe().mean()
kids_summary = df[['Kidhome', 'Teenhome']].describe().mean()

print("Resumen estadístico de Ingresos:")
print(income_summary)

print("\nResumen estadístico de Edad:")
print(age_summary)

print("\nResumen estadístico de Cantidad Gastada en Productos:")
print(spent_summary)

print("\nResumen estadístico de Número de Hijos:")
print(kids_summary)
```

```python
# Varianza y desviación estándar de Ingresos
var_income = df['Income'].var()
std_income = df['Income'].std()


print(f"Varianza de Ingresos: {var_income}")
print(f"Desviación estándar de Ingresos: {std_income}")
```

-la varianza de ingresos muestra una gran dispersion entre los ingresos mas bajos y los mas altos lo que se ve la diversidad de ingresos entre los clientes.
-La desviación estándar de los ingresos también es significativa  lo que confirma que los ingresos de los clientes están muy alejados de la media

```python
import matplotlib.pyplot as plt

# Sumar las columnas de gasto en productos
df['TotalSpent'] = df['MntWines'] + df['MntFruits'] + df['MntMeatProducts'] + df['MntFishProducts'] + df['MntSweetProducts'] + df['MntGoldProds']

# Crear el diagrama de dispersión entre Ingresos y Total gastado
plt.figure(figsize=(8, 6))
plt.scatter(df['Income'], df['TotalSpent'], alpha=0.5, c='green', edgecolors='black')
plt.title('Diagrama de Dispersión: Ingresos vs Gasto Total en Productos')
plt.xlabel('Ingresos')
plt.ylabel('Total Gastado en Productos')
plt.grid(True)
plt.show()
```

El grafico muestra que esta de forma ascendente lo que significa que mientras los clientes tengan mas ingresos gastan mas en productos
